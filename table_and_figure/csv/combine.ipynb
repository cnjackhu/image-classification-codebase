##### because the  cifar100_lamb_wd file don't have baseline result, we need add the baseline(reg=0) results to the csv file

```python
import pandas as pd

# 1. Read the CSV files
df1 = pd.read_csv('wandb_cifar10.csv')
df2 = pd.read_csv('wandb_cifar10_lamb_wd.csv')
x = df1[(df1["reg"] == 0) & (df1["celoss"] == 1)]
combined_df = pd.concat([df2 , x], ignore_index=True)
combined_df.to_csv('wandb_cifar10_lamb_wd.csv', index=False)
```

```python
# 1. Read the CSV files
df1 = pd.read_csv('test_cifar10+celoss_combined.csv')
df2 = pd.read_csv('test_cifar10_lamb_wd.csv')

x = df1[(df1["reg"] == 0) & (df1["celoss"] == 1)]
combined_df = pd.concat([df2 , x], ignore_index=True)
combined_df.to_csv('test_cifar10_lamb_wd.csv', index=False)
```
