```python
import pandas as pd

# Load the CSV file you saved earlier
df = pd.read_csv("L.csv")
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", palette="colorblind", font_scale=1.5)
plt.figure(figsize=(10, 6))

# Melt the DataFrame if it has multiple time series columns
df_melted = df.melt(id_vars=["step"], var_name="Series", value_name="Value")

sns.lineplot(data=df_melted, hue="Series")

plt.title("Time Series Comparison")
plt.xlabel("Step")
plt.ylabel("Metric Value")
plt.legend(title="Run")
plt.tight_layout()
plt.savefig("comparison_plot.pdf")
plt.show()


```
